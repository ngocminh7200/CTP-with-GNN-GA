```python
#To randomly assign classroom and time slots to a course

C = len(course_list)  # number of courses
T = 25                # timeslots
R = len(room_list)    # number of rooms

#given a course list (keys : prof, program, numstu)
def random_individual():
    timeslot_genes = np.random.randint(0, T, size=C)  # random timeslot for each course
    room_genes     = np.random.randint(0, R, size=C)  # random room for each course
    return np.stack([timeslot_genes, room_genes], axis=1)  # shape (C, 2)

    
ind = random_individual()
print(ind)
print(f"Individual shape: {ind.shape}")
print(f"Sample gene 0: timeslot={timeslots[ind[0,0]]}, room={room_list[ind[0,1]]}")
print(f"Sample gene 1: timeslot={timeslots[ind[1,0]]}, room={room_list[ind[1,1]]}")

#a sample gene is a course 
#an allel is the course allocation
```

```python
#To convert a chromosome/schedule into a dataframe suitable for graph input

def decode_individual(ind):
    df = pd.DataFrame({
        'course index': np.arange(C),  #enumaerate the keys/course list
        'Key':          course_list,   
        'timeslot':     ind[:, 0] + 1,          # 1-indexed to match GNN
        'timeslot_label': [timeslots[i] for i in ind[:, 0]],  # readable label e.g. 'MT3'
        'classroom index': ind[:, 1],
        'room':         [room_list[i] for i in ind[:, 1]],    # readable label e.g. 'FAW.1.01'
        'capacity':     [room_capacities[i] for i in ind[:, 1]],
        'professor':    schedule_df['Staff'].tolist(),
        'numstu':       schedule_df['NUMSTU'].tolist(),
        'dl':           schedule_df['Program'].tolist(),
    })
    return df
    
decode_individual(ind)
```

```python
# a course cannot be given two time slots 
# a room at the same times can be given to two different courses  - hard contraint
#a proof teaching two keys is scheduled at the same time slot -  hard contraint
# capacity violation -  room capacity smaller than numstudents

#keys with same programme are scheduled at the same time slots - we want minimal overlapping - soft constaint 
```

```python
# professor conflict
# The different courses taught by the same prof are given in the same timeslot.
#professor double booking


def count_professor_conflicts(schedule):
    # group by professor AND timeslot, count courses in each group
    prof_ts = schedule.groupby(['professor', 'timeslot']).size()
    
    # any group with more than 1 course = conflict
    conflicts = prof_ts[prof_ts > 1]
    
    # total number of conflicts
    H2 = (prof_ts > 1).sum()
    
    return H2, conflicts


a,b=count_professor_conflicts(chromosome1)
print(a)
print(b)
```

```python
# room conflict
# the same room is being used by different courses at the same timeslot
# room double booking

def count_room_conflicts(schedule):
    # group by room AND timeslot, count courses in each group
    room_ts = schedule.groupby(['classroom index', 'timeslot']).size()
    
    # any group with more than 1 course = conflict
    conflicts = room_ts[room_ts > 1]
    
    # total number of conflicts
    H3 = (room_ts > 1).sum()
    
    return H3, conflicts

a,b=count_room_conflicts(chromosome1)
print(a)
print(b)
```

```python
# room capacity violation
# the number of students in a course exceeds the room capacity
# H4


def count_capacity_violations(schedule):
    # directly compare numstu vs capacity for each course
    violations = schedule[schedule['numstu'] > schedule['capacity']]
    
    # total number of violations
    H4 = (schedule['numstu'] > schedule['capacity']).sum()
    
    return H4, violations
    
a,b=count_capacity_violations(chromosome1)
print(a)
print(b)
```

```python
# program overlap
# students in the same program have 2+ courses at the same timeslot
# S2 - SOFT constraint
def count_program_overlaps(schedule):
    # group by program AND timeslot, count courses in each group
    prog_ts = schedule.groupby(['dl', 'timeslot']).size()
    
    # any group with more than 1 course = overlap
    conflicts = prog_ts[prog_ts > 1]
    
    # total number of overlaps
    S2 = (prog_ts > 1).sum()
    
    return S2, conflicts

a,b=count_program_overlaps(chromosome1)
print(a)
print(b)
```

```python
def fitness_function(schedule):
    # hard constraints (violations)
    H2, _ = count_professor_conflicts(schedule)
    H3, _ = count_room_conflicts(schedule)
    H4, _ = count_capacity_violations(schedule)
    
    # soft constraint (violations)
    S2, _ = count_program_overlaps(schedule)
    
    # weights
    w_H2 = 10   # professor conflict
    w_H3 = 10   # room conflict
    w_H4 = 5    # capacity violation
    w_S2 = 1    # program overlap
    
    # soft constraint score (positive)
    soft_score = w_S2 * S2
    
    # hard constraint penalty (squared → always hurts a lot)
    hard_penalty = (w_H2 * H2) + (w_H3 * H3) + (w_H4 * H4)
    
    # fitness function from paper
    fitness = soft_score + (hard_penalty ** 2)
    #make it as close to 0 as possible
    #if fitness is negative, it means that the chromosome is not valid(hard constraint is violated)


    
    return int(fitness), int(H2), int(H3), int(H4), int(S2)
print(fitness_function(chromosome1))
```

```python
population_fitness = []
for ind in population:
    schedule = decode_individual(ind)            # decode outside
    fitness, H2, H3, H4, S2 = fitness_function(schedule)  # pass decoded df
    population_fitness.append([fitness, H2, H3, H4, S2])

fitness_df = pd.DataFrame(population_fitness, columns=['fitness', 'H2', 'H3', 'H4', 'S2'])
print(fitness_df)

scores = fitness_df['fitness'].tolist()
print(scores[:5])  # first 5 scores
```

```python
#Pick k random individuals, return the best one scores = list of fitness values
#(higher is better as our fitness function gives negative values)
def tournament_selection(population, scores, k=3):

    candidates = random.sample(range(len(population)), k)
    best = max(candidates, key=lambda x: scores[x])
    return population[best]

#to keep top elite individuals unchanged
def get_elite(population, scores, elite_size=2):

    # sort by fitness descending (highest first)
    elite_indices = sorted(range(len(population)), key=lambda x: scores[x], reverse=True)[:elite_size]
    return [population[i] for i in elite_indices]
```

```python
#TYPES OF CROSSOVERS
#1. One Point Crossover
#2. Multi Point Crossover
#3. Multi-point Class Crossover
#4. Multi-point Student Crossover

# find where each program starts and ends in course_df
program_boundaries = []
programs = course_df['Program'].tolist()

current_program = programs[0]
start_idx = 0

for i, prog in enumerate(programs):
    if prog != current_program:
        program_boundaries.append((current_program, start_idx, i))
        current_program = prog
        start_idx = i

# add last program
program_boundaries.append((current_program, start_idx, len(programs)))

print(f"Number of programs: {len(program_boundaries)}")
print(f"First 5 boundaries: {program_boundaries[:5]}")
```

```python
def crossover(parent1, parent2, program_boundaries, num_points=3):

    child = np.zeros_like(parent1)
    
    # pick random program split points
    split_indices = sorted(random.sample(range(len(program_boundaries)), num_points)) #choose rnaodm splits
    
    # alternate between parent1 and parent2 at each split
    current_parent = parent1
    prev_end = 0
    
    for i, idx in enumerate(split_indices):
        prog, start, end = program_boundaries[idx]
        
        # fill from prev_end to start of this program from current parent
        child[prev_end:start] = current_parent[prev_end:start]
        
        # switch parent at split point
        current_parent = parent2 if current_parent is parent1 else parent1
        prev_end = start
    
    # fill remainder from current parent
    child[prev_end:] = current_parent[prev_end:]
    
    return child
```
